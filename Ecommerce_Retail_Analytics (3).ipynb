```python
file_path = "/Workspace/Users/surbhiaglawe143@gmail.com/online_retail.csv"

sales_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("timestampFormat", "yyyy-MM-dd HH:mm:ss")
    .csv(file_path)
)

print("Total rows:", sales_df.count())
print("Total columns:", len(sales_df.columns))

sales_df.printSchema()
display(sales_df.limit(10))
```

```python
from pyspark.sql.functions import (
    col,
    trim,
    when,
    count,
    sum as spark_sum
)

print("========== DATA QUALITY REPORT ==========")

print("Total rows:", sales_df.count())

print("Duplicate rows:", 
      sales_df.count() - sales_df.dropDuplicates().count())

print("\nMissing values by column:")

sales_df.select([
    spark_sum(
        when(
            col(c).isNull() |
            (trim(col(c)) == "" if c in ["Description", "CustomerID", "Country"] else col(c).isNull()),
            1
        ).otherwise(0)
    ).alias(c)
    for c in sales_df.columns
]).show(truncate=False)
```

```python
from pyspark.sql.functions import (
    col,
    trim,
    when,
    lit,
    to_timestamp,
    round as spark_round
)

# 1. Remove exact duplicate rows
clean_df = sales_df.dropDuplicates()

# 2. Clean text columns
for column_name in ["Description", "CustomerID", "Country"]:
    clean_df = clean_df.withColumn(
        column_name,
        when(
            col(column_name).isNull() |
            (trim(col(column_name)) == ""),
            lit(None).cast("string")
        ).otherwise(trim(col(column_name)))
    )

# 3. Add data-quality flags
clean_df = clean_df.withColumn(
    "MissingDescription",
    col("Description").isNull()
)

clean_df = clean_df.withColumn(
    "MissingCustomerID",
    col("CustomerID").isNull()
)

clean_df = clean_df.withColumn(
    "NegativeQuantity",
    col("Quantity") < 0
)

clean_df = clean_df.withColumn(
    "InvalidPrice",
    col("UnitPrice").isNull() | (col("UnitPrice") < 0)
)

clean_df = clean_df.withColumn(
    "ZeroPrice",
    col("UnitPrice") == 0
)

# 4. Calculate revenue
clean_df = clean_df.withColumn(
    "Revenue",
    when(
        col("Quantity").isNotNull() &
        col("UnitPrice").isNotNull(),
        spark_round(col("Quantity") * col("UnitPrice"), 2)
    ).otherwise(lit(None).cast("double"))
)

# 5. Verify the result
print("Original rows:", sales_df.count())
print("Cleaned rows:", clean_df.count())
print("Rows removed:", sales_df.count() - clean_df.count())

display(clean_df.limit(10))
```

```python
# Save the cleaned DataFrame as a managed Delta table
clean_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("cleaned_online_retail")

print("Cleaned dataset saved successfully!")
```

```python
from pyspark.sql.functions import col, trim, when, lit, round as spark_round

def clean_retail_data(raw_df):
    # Remove exact duplicate rows
    df = raw_df.dropDuplicates()

    # Normalize text columns
    for name in ["Description", "CustomerID", "Country"]:
        df = df.withColumn(
            name,
            when(
                col(name).isNull() | (trim(col(name)) == ""),
                lit(None).cast("string")
            ).otherwise(trim(col(name)))
        )

    # Add data-quality flags
    df = df.withColumn(
        "MissingDescription", col("Description").isNull()
    )
    df = df.withColumn(
        "MissingCustomerID", col("CustomerID").isNull()
    )
    df = df.withColumn(
        "NegativeQuantity", col("Quantity") < 0
    )
    df = df.withColumn(
        "InvalidPrice",
        col("UnitPrice").isNull() | (col("UnitPrice") < 0)
    )
    df = df.withColumn(
        "ZeroPrice", col("UnitPrice") == 0
    )

    # Calculate revenue
    df = df.withColumn(
        "Revenue",
        when(
            col("Quantity").isNotNull() & col("UnitPrice").isNotNull(),
            spark_round(col("Quantity") * col("UnitPrice"), 2)
        ).otherwise(lit(None).cast("double"))
    )

    return df

print("Cleaning function created successfully!")
```

```python
from pyspark.sql.functions import col

def validate_and_save(clean_df):
    print("========== VALIDATION ==========")

    row_count = clean_df.count()

    duplicate_count = (
        row_count - clean_df.dropDuplicates().count()
    )

    missing_invoice_count = clean_df.filter(
        col("InvoiceNo").isNull()
    ).count()

    missing_date_count = clean_df.filter(
        col("InvoiceDate").isNull()
    ).count()

    incorrect_revenue_count = clean_df.filter(
        col("Revenue").isNull()
        | (
            col("Revenue")
            != (col("Quantity") * col("UnitPrice")).cast("decimal(20,2)")
        )
    ).count()

    print("Cleaned rows:", row_count)
    print("Remaining duplicates:", duplicate_count)
    print("Missing invoice numbers:", missing_invoice_count)
    print("Missing invoice dates:", missing_date_count)
    print("Incorrect revenue rows:", incorrect_revenue_count)

    if duplicate_count != 0:
        raise ValueError("Duplicate rows found!")

    if missing_invoice_count != 0:
        raise ValueError("Missing invoice numbers found!")

    if missing_date_count != 0:
        raise ValueError("Missing invoice dates found!")

    if incorrect_revenue_count != 0:
        raise ValueError("Incorrect revenue calculations found!")

    clean_df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable("cleaned_online_retail")

    print("Validation passed!")
    print("Cleaned data saved successfully!")
```
